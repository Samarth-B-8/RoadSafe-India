# RoadSafe India — Fatality Analysis

## Objective

This notebook examines road-accident fatalities across Indian States/UTs
from 2020 to 2024.

The analysis will investigate:

1. Fatality trends across the study period.
2. States/UTs with the highest fatality counts.
3. Changes in fatalities between 2020 and 2024.
4. Fatalities relative to the number of reported accidents.
5. Fatalities per 100,000 projected population.

The fatality dataset is analyzed separately before being integrated with
the accident dataset.

```python
import pandas as pd
```

```python
fatality_path = (
    "../data/raw/"
    "state_wise_road_fatalities_2020_2024.csv"
)

fatality_df = pd.read_csv(fatality_path)

print("Fatality dataset loaded successfully.")
print(f"Rows: {fatality_df.shape[0]}")
print(f"Columns: {fatality_df.shape[1]}")
```

```python
print("First 5 rows:")
display(fatality_df.head())

print("\nLast 5 rows:")
display(fatality_df.tail())
```

```python
print("Column names:")
for i, column in enumerate(fatality_df.columns, start=1):
    print(f"{i}. {column}")

print("\nData types:")
display(fatality_df.dtypes)

print("\nDataset information:")
fatality_df.info()
```

```python
print("Missing values:")
display(fatality_df.isnull().sum())

print(
    "\nTotal missing values:",
    fatality_df.isnull().sum().sum()
)
```

```python
print("Rows with missing State/UT:")

missing_state_rows = fatality_df[
    fatality_df["State"].isna()
]

display(missing_state_rows)
```

```python
print("State/UT values:")

display(
    fatality_df["State"]
    .dropna()
    .unique()
)
```

```python
# Standardize Column Names
fatality_df.columns = (
    fatality_df.columns
    .str.strip()
    .str.lower()
    .str.replace("%", "percent", regex=False)
    .str.replace(" ", "_", regex=False)
    .str.replace("-", "_", regex=False)
)

print("Standardized columns:")
print(fatality_df.columns.tolist())
```

```python
fatality_clean = fatality_df.copy()

print("Working copy created.")
```

```python
fatality_clean["state"] = (
    fatality_clean["state"]
    .astype("string")
    .str.strip()
)

print("State/UT values cleaned.")
```

```python
killed_columns = [
    "2020_killed",
    "2021_killed",
    "2022_killed",
    "2023_killed",
    "2024_killed"
]

for column in killed_columns:
    fatality_clean[column] = (
        fatality_clean[column]
        .astype("string")
        .str.replace(",", "", regex=False)
        .str.strip()
    )

    fatality_clean[column] = pd.to_numeric(
        fatality_clean[column],
        errors="coerce"
    )

print("Killed columns converted to numeric.")
```

```python
invalid_fatality_rows = fatality_clean[
    fatality_clean["state"].isna() |
    fatality_clean["2020_killed"].isna()
].copy()

print("Potentially invalid fatality rows:")
display(invalid_fatality_rows)
```

```python
fatality_clean = fatality_clean[
    fatality_clean["state"].notna()
].copy()

fatality_clean = fatality_clean[
    fatality_clean[killed_columns].notna().any(axis=1)
].copy()

print(
    "Rows remaining after removing invalid records:",
    len(fatality_clean)
)
```

```python
fatality_clean = fatality_clean[
    ~fatality_clean["state"].str.contains(
        "total|india",
        case=False,
        na=False
    )
].copy()

print(
    "State/UT records remaining:",
    len(fatality_clean)
)
```

```python
print(
    "Duplicate complete rows:",
    fatality_clean.duplicated().sum()
)

print(
    "Duplicate State/UT names:",
    fatality_clean["state"].duplicated().sum()
)

if fatality_clean["state"].duplicated().any():
    print("\nDuplicated State/UT entries:")
    display(
        fatality_clean[
            fatality_clean["state"].duplicated(keep=False)
        ].sort_values("state")
    )
```

```python
print("Minimum fatality count by year:")
display(
    fatality_clean[killed_columns].min()
)

negative_rows = fatality_clean[
    (fatality_clean[killed_columns] < 0).any(axis=1)
]

print("\nRows with negative fatality values:")

if negative_rows.empty:
    print("No negative fatality values found.")
else:
    display(negative_rows)
```

```python
print("Cleaned fatality dataset:")

display(
    fatality_clean[
        [
            "state",
            "2020_killed",
            "2021_killed",
            "2022_killed",
            "2023_killed",
            "2024_killed"
        ]
    ].head(10)
)
```

```python
print("FATALITY CLEANING SUMMARY")
print("-------------------------")

print("Rows:", fatality_clean.shape[0])
print("Columns:", fatality_clean.shape[1])

print(
    "Missing values:",
    fatality_clean.isnull().sum().sum()
)

print(
    "Duplicate State/UTs:",
    fatality_clean["state"].duplicated().sum()
)

print(
    "Unique State/UTs:",
    fatality_clean["state"].nunique()
)
```

```python
# Identify exactly where the missing values are

missing_summary = fatality_clean.isnull().sum()

print("Missing values by column:")
display(
    missing_summary[missing_summary > 0]
)

print("\nRows containing missing values:")
display(
    fatality_clean[
        fatality_clean.isnull().any(axis=1)
    ]
)
```

```python
# Identify the State/UT responsible for the missing values

missing_rows = fatality_clean[
    fatality_clean[
        [
            "2020_killed",
            "2020_ranking",
            "2021_ranking"
        ]
    ].isnull().any(axis=1)
]
    
display(missing_rows)
```

```python
fatality_output_path = (
    "../data/processed/"
    "state_wise_road_fatalities_2020_2024_cleaned.csv"
)

fatality_clean.to_csv(
    fatality_output_path,
    index=False
)

print("Cleaned fatality dataset saved:")
print(fatality_output_path)
```

## Initial Fatality Data Assessment

The State/UT-level fatality dataset contains annual reported fatalities from
2020 to 2024 along with year-over-year change and ranking information.

The original fatality CSV was preserved in `data/raw/`.

The cleaned dataset was saved in `data/processed/`.

Three missing values were identified for Ladakh. The source reports the 2020
fatality value for Ladakh as "NA", while fatalities are reported for 2021,
2022, 2023 and 2024. Therefore, the missing 2020 value is retained as missing
rather than being imputed as zero.

Ladakh will be excluded from analyses requiring a 2020 fatality value but will
be included in analyses for years where its fatality data is available.

The cleaned fatality data will be used for subsequent analysis of:
- State/UT fatality burden.
- Fatality trends.
- Fatalities per 100,000 projected population.
- Fatalities relative to reported accidents.
