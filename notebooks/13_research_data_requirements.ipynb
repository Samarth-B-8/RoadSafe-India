## 6. Hypothesized Relationships
### H1 — Vehicle Exposure

States/UTs with higher vehicle ownership are expected to have higher
reported accident and fatality burdens.

### H2 — Road Infrastructure

Differences in road infrastructure are associated with differences in
reported accident and fatality outcomes.

### H3 — Population Density

States/UTs with greater population density may exhibit different accident
and fatality patterns from less densely populated States/UTs.

### H4 — Socioeconomic Context

Socioeconomic characteristics may be associated with State/UT-level
differences in reported road-safety outcomes.

### H5 — Multi-factor Relationship

Accident and fatality outcomes are likely to be associated with a
combination of exposure, infrastructure and socioeconomic characteristics
rather than a single factor.

## 13. Vehicle Exposure Data

### Candidate Variable

The first additional explanatory variable will be State/UT-level vehicle
registrations.

Candidate variables include:

- Total vehicle registrations
- Vehicles per 1,000 population
- Vehicle registration growth

### Research Motivation

Population is an incomplete proxy for road-traffic exposure. Vehicle
registrations provide an additional State/UT-level indicator of the size of
the motor-vehicle fleet.

### Primary Candidate Source

Vahan Public Dashboard — Ministry of Road Transport & Highways,
Government of India.

The Vahan dashboard provides State-level filtering and historical vehicle
registration information.

### Important Limitation

Registered vehicles are not equivalent to actual road usage or
vehicle-kilometres travelled. Therefore, vehicle registrations will be
treated as an exposure proxy rather than a direct measure of traffic
exposure.

## 14. Road Infrastructure Data

### Candidate Variables

The first infrastructure variables to investigate are:

- Total road length
- National Highway length
- State Highway length

### Primary Research Variable

The initial variable will be:

- Total road length in kilometres

### Derived Indicator

Road length per 1,000 projected population.

### Research Motivation

Road infrastructure represents an important structural characteristic of
a State/UT. Differences in road-network size may be associated with differences
in reported accident and fatality burden.

### Candidate Sources

The Reserve Bank of India's Handbook of Statistics on Indian States contains
State-wise road infrastructure tables, including:

- Table 141 — State-wise Length of National Highways
- Table 143 — State-wise Length of Roads
- Table 144 — State-wise Length of State Highways

The Ministry of Road Transport & Highways also publishes official
State-wise National Highway information.

### Important Limitation

Road length alone does not measure road quality, traffic volume, congestion,
vehicle-kilometres travelled, lane capacity or road-user exposure.

Therefore, road length will be treated as an infrastructure indicator rather
than a direct measure of road-safety risk.
