# RoadSafe India — Data Understanding

## Objective

This notebook is the first analytical stage of the RoadSafe India project.

The goal is to understand the available road-accident dataset before performing
cleaning, exploratory data analysis, statistical analysis, visualization, or
machine learning.

## Central Research Question

How do road accident frequency, severity and related factors vary across India,
across time and across different road/environmental conditions, and what
patterns can be identified that may help prioritize road-safety interventions?

## Notebook Goals

1. Understand the dataset and its structure.
2. Identify the meaning and type of each variable.
3. Assess data quality and missing values.
4. Identify potential limitations and biases.
5. Determine what analytical questions the dataset can realistically support.

```python
import pandas as pd
```

```python
df = pd.read_csv("D:/Data Science/Projects/RoadSafe-India/data/raw/state_wise_road_accidents_2020_2024.csv")
```

```python
# Display the first 5 rows
print("First 5 rows:")
display(df.head())

# Display the last 5 rows
print("\nLast 5 rows:")
display(df.tail())
```

```python
# Number of rows and columns
print("Dataset shape:")
print(df.shape)

print(f"\nNumber of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")
```

```python
# Display all column names
print("Column names:")
for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")
```

```python
# Display all column names
print("Column names:")
for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")
```

```python
# Display the data type of every column
print("Data types:")
display(df.dtypes)
```

```python
# Display structural information about the dataset
print("Dataset information:")
df.info()
```

```python
# Count missing values in every column
missing_values = df.isnull().sum()

print("Missing values in each column:")
display(missing_values)

print(f"\nTotal missing values in the dataset: {missing_values.sum()}")
```

```python
# Check for completely duplicated rows
duplicate_rows = df.duplicated().sum()

print(f"Number of duplicate rows: {duplicate_rows}")
```

```python
# Check the number of unique State/UT entries
unique_states = df["State"].nunique()

print(f"Number of unique State/UT names: {unique_states}")

print("\nDuplicate State/UT names:")
display(df[df["State"].duplicated(keep=False)].sort_values("State"))
```

```python
# Statistical summary of numerical columns
print("Statistical summary of numerical columns:")
display(df.describe())
```

```python
# Identify columns that Pandas recognizes as numerical
numeric_columns = df.select_dtypes(include="number").columns

print("Numerical columns:")
for column in numeric_columns:
    print("-", column)
```

```python
# Negative accident counts/rankings would be suspicious,
# so we perform a basic sanity check.

minimum_values = df[numeric_columns].min()

print("Minimum value in each numerical column:")
display(minimum_values)

print("\nColumns containing negative values:")

negative_columns = minimum_values[minimum_values < 0]

if negative_columns.empty:
    print("No numerical column contains negative values.")
else:
    display(negative_columns)
```

```python
# Display all State/UT names
print("State/UT entries:")
display(df["State"].tolist())
```

```python
# Confirm whether each row corresponds to a unique State/UT
state_counts = df["State"].value_counts()

print("State/UT frequency:")
display(state_counts)

print("\nStates/UTs appearing more than once:")
display(state_counts[state_counts > 1])
```

```python
# Identify the accident-count columns
accident_columns = [
    column
    for column in df.columns
    if "Accidents" in column
]

print("Accident-count columns:")
for column in accident_columns:
    print("-", column)

print("\nMinimum accident count:")
display(df[accident_columns].min())

print("\nMaximum accident count:")
display(df[accident_columns].max())
```

```python
# Compare accident counts across the five years
year_columns = [
    "2020 Accidents",
    "2021 Accidents",
    "2022 Accidents",
    "2023 Accidents",
    "2024 Accidents"
]

print("Accident counts from 2020 to 2024:")
display(df[["State"] + year_columns])
```

```python
print("Initial Data Understanding Observations")
print("--------------------------------------")

print(f"1. The dataset contains {df.shape[0]} rows and {df.shape[1]} columns.")

print(
    f"2. The dataset contains {df['State'].nunique()} unique State/UT entries."
)

print(
    f"3. The dataset contains {missing_values.sum()} missing values in total."
)

print(
    f"4. The dataset contains {duplicate_rows} completely duplicated rows."
)

print(
    "5. Each row represents an aggregated State/UT-level observation, "
    "rather than an individual road accident."
)

print(
    "6. The dataset contains yearly road-accident counts for 2020, 2021, "
    "2022, 2023 and 2024."
)

print(
    "7. The dataset also contains year-over-year change information "
    "and State/UT rankings."
)

print(
    "8. This dataset is therefore suitable for state-level trend, "
    "comparison and ranking analysis."
)

print(
    "9. This dataset does not provide individual accident-level "
    "characteristics such as accident time, weather, lighting or "
    "individual accident severity."
)
```
